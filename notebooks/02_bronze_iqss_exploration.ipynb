# Bronze IQSS Data Exploration

This notebook explores the **raw IQSS data** from the Bronze layer.

IQSS = Indicateurs de Qualité et de Sécurité des Soins (Quality and Safety Indicators)

```python
import sys
from pathlib import Path

import pandas as pd

sys.path.append('..')

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)
```

## 1. Select Year and Load Data

Choose which year to explore (2019, 2023, etc.)

```python
# Configuration
YEAR = 2023  # Change this to explore different years

# Find the file
bronze_dir = Path('../data/bronze/iqss') / str(YEAR)
files = list(bronze_dir.glob('resultats_iqss_*.csv')) + list(bronze_dir.glob('resultats_iqss_*.xlsx'))

if not files:
    print(f"❌ No IQSS data found for year {YEAR}")
    print(f"   Run: python ../main.py --years {YEAR}")
else:
    filepath = files[0]
    print(f"📁 Found: {filepath.name}")
    print(f"📏 Size: {filepath.stat().st_size / 1024**2:.2f} MB")
```

```python
# Load the data
ext = filepath.suffix.lower()

if ext == '.csv':
    df_bronze = pd.read_csv(filepath, sep=';', dtype=str, encoding='latin-1')
elif ext in ['.xlsx', '.xls']:
    df_bronze = pd.read_excel(filepath, dtype=str)
else:
    raise ValueError(f"Unsupported file format: {ext}")

print(f"\n📊 Shape: {df_bronze.shape}")
print(f"\n📋 Columns ({len(df_bronze.columns)}):")
for i, col in enumerate(df_bronze.columns, 1):
    print(f"  {i:2}. {col}")
```

## 2. Data Overview

```python
# Basic info
print("📈 Dataset Statistics:")
print(f"  Total rows: {len(df_bronze):,}")
print(f"  Total columns: {len(df_bronze.columns)}")
print(f"  Memory usage: {df_bronze.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# Display first few rows
print("\n📋 First 5 Rows:")
df_bronze.head()
```

## 3. Column Analysis

```python
# Categorize columns by type
nb_cols = [c for c in df_bronze.columns if 'nb_' in c.lower()]
score_cols = [c for c in df_bronze.columns if 'score' in c.lower()]
categorical_cols = [c for c in df_bronze.columns if c.lower() in ['participation', 'depot']]

print("📊 Column Categories:")
print(f"  Count columns (nb_*): {len(nb_cols)}")
print(f"  Score columns: {len(score_cols)}")
print(f"  Categorical: {len(categorical_cols)}")
print(f"  Other: {len(df_bronze.columns) - len(nb_cols) - len(score_cols) - len(categorical_cols)}")
```

```python
# Null value analysis
null_counts = df_bronze.isnull().sum()
null_pct = (null_counts / len(df_bronze) * 100).round(2)

null_summary = pd.DataFrame({
    'Column': df_bronze.columns,
    'Null_Count': null_counts.values,
    'Null_Percentage': null_pct.values
}).sort_values('Null_Percentage', ascending=False)

print("🔍 Columns with >50% Null Values:")
print(null_summary[null_summary['Null_Percentage'] > 50])
```

## 4. Categorical Values (Before Cleaning)

```python
# Check participation and depot columns for numbered prefixes
for col in categorical_cols:
    if col in df_bronze.columns:
        print(f"\n📊 {col.upper()} Values:")
        print(df_bronze[col].value_counts())
```
