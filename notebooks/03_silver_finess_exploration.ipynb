# Silver FINESS Data Exploration

This notebook explores the **cleaned FINESS data** from the Silver layer.

This data has been processed by `transform/finess_cleaner.py`.

```python
import pandas as pd
import numpy as np
import sys
sys.path.append('..')

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)
```

## 1. Load Cleaned FINESS Data

```python
# Load silver/cleaned data
df_silver = pd.read_csv('../data/silver/finess/finess.csv', sep=';')

print(f"📊 Shape: {df_silver.shape}")
print(f"\n📋 Columns ({len(df_silver.columns)}):")
for i, col in enumerate(df_silver.columns, 1):
    print(f"  {i:2}. {col}")
```

## 2. Data Quality Overview

```python
# Basic statistics
print("📈 Dataset Statistics:")
print(f"  Total establishments: {len(df_silver):,}")
print(f"  Total columns: {len(df_silver.columns)}")
print(f"  Memory usage: {df_silver.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# Data types
print(f"\n📝 Data Types:")
print(df_silver.dtypes.value_counts())
```

```python
# Display first few rows
print("📋 First 5 Rows:")
df_silver.head()
```

## 3. Data Completeness

```python
# Null value analysis
null_counts = df_silver.isnull().sum()
null_pct = (null_counts / len(df_silver) * 100).round(2)

completeness = pd.DataFrame({
    'Column': df_silver.columns,
    'Non_Null_Count': (len(df_silver) - null_counts).values,
    'Null_Count': null_counts.values,
    'Completeness_%': (100 - null_pct).values
}).sort_values('Completeness_%')

print("🔍 Data Completeness:")
print(completeness)
```

## 4. Cleaned Field Examples

```python
# Check FINESS identifier formatting (should be 9 digits)
if 'finess' in df_silver.columns:
    print("🔢 FINESS Identifier:")
    print(f"  Min length: {df_silver['finess'].str.len().min()}")
    print(f"  Max length: {df_silver['finess'].str.len().max()}")
    print(f"  Sample: {df_silver['finess'].head(5).tolist()}")

# Check SIRET formatting (should be 14 digits)
if 'siret' in df_silver.columns:
    print("\n🔢 SIRET:")
    print(f"  Min length: {df_silver['siret'].str.len().min()}")
    print(f"  Max length: {df_silver['siret'].str.len().max()}")
    print(f"  Sample: {df_silver['siret'].head(5).tolist()}")
```

```python
# Check text field cleaning (should be title case)
if 'rs' in df_silver.columns:
    print("✨ Name Cleaning Examples:")
    print(df_silver[['finess', 'rs', 'commune']].head(10))
```

## 5. Distribution Analysis

```python
# Regional distribution
if 'libregion' in df_silver.columns:
    print("🗺️ Top 10 Regions by Establishment Count:")
    print(df_silver['libregion'].value_counts().head(10))
```

```python
# Establishment type distribution
if 'libmft' in df_silver.columns:
    print("\n🏥 Top 10 Establishment Types:")
    print(df_silver['libmft'].value_counts().head(10))
```

## 6. Sample Records

```python
# Random sample for inspection
print("🏥 Sample Establishments (Random 3):")
df_silver.sample(min(3, len(df_silver))).T
```

## 7. Data Quality Checks

```python
# Check for duplicates on FINESS identifier
if 'finess' in df_silver.columns:
    duplicate_count = df_silver['finess'].duplicated().sum()
    print(f"🔍 Quality Checks:")
    print(f"  Duplicate FINESS IDs: {duplicate_count}")
    
    if duplicate_count > 0:
        print("\n  Duplicate IDs:")
        duplicates = df_silver[df_silver['finess'].duplicated(keep=False)].sort_values('finess')
        print(duplicates[['finess', 'rs', 'commune']].head(10))
```
